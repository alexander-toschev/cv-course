## Паспорт массива / Array properties

Назовите смысл каждого размера в shape. / Explain each dimension in the shape.


```python
import numpy as np
a = np.arange(6).reshape(2, 3)
print(a, a.shape, a.ndim, a.dtype, sep="\n")
```

## Индексы и срезы / Indices and slices

Здесь числа начинаются с 0, в Task0 — с 1. / Values start at 0 here and at 1 in Task0.


```python
a = np.arange(24).reshape(4, 6)
print(a[1, ::-1])
assert np.array_equal(a[1, ::-1], [11, 10, 9, 8, 7, 6])
```

## View и copy / View and copy

Объясните, какое присваивание меняет a. / Explain which assignment changes a.


```python
a = np.arange(4)
v = a[:2]
v[0] = 99
assert a[0] == 99
c = a[:2].copy()
c[0] = -1
assert a[0] == 99
print(a, c)
```

## Агрегация / Reductions

Предскажите форму каждого результата. / Predict the shape of each result.


```python
a = np.arange(6).reshape(2, 3)
print(a.mean(), a.mean(axis=0), a.mean(axis=1, keepdims=True), sep="\n")
```

## Broadcasting / Broadcasting

Сравните размеры справа налево. / Compare dimensions from right to left.


```python
x = np.arange(3).reshape(3, 1)
y = np.arange(4).reshape(1, 4)
print(x + y)
assert (x + y).shape == (3, 4)
```

## Маска / Boolean masks

Почему пустая выборка не равна нулевому среднему? / Why does an empty selection not mean a zero average?


```python
z = np.array([1., 2., 5., 8.])
selected = z[z > 3]
print(selected, selected.mean())
empty = z[z > 100]
print(empty.mean() if empty.size else np.nan)
```

## Переполнение / Overflow

Сравните uint8 и float32 до сложения. / Compare uint8 and float32 before addition.


```python
a = np.array([250], dtype=np.uint8)
b = np.array([10], dtype=np.uint8)
print(a + b, a.astype(np.float32) + b)
assert int((a + b)[0]) == 4
```

## RGB-пример / RGB example

Проверьте неизменность исходного изображения. / Verify that the source image stays unchanged.


```python
img = np.arange(72, dtype=np.uint8).reshape(4, 6, 3)
original = img.copy()
f = img.astype(np.float32) / np.float32(255)
roi = f[1:3, 1:5, :].copy()
gain = np.array([1.2, 1.0, 0.8], dtype=np.float32)
corrected = np.clip(roi * gain, 0, 1)
print(corrected.shape, corrected.dtype, corrected.mean(axis=(0, 1)))
assert np.array_equal(img, original)
assert corrected.shape == (2, 4, 3)
```

## Решение СЛАУ / Solving a linear system

Объясните невязку A @ x - b. / Explain the residual A @ x - b.


```python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([4., 7.])
x = np.linalg.solve(A, b)
print(x, A @ x - b)
assert np.allclose(A @ x, b)
```
