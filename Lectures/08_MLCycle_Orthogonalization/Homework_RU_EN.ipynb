```python
# Не редактируйте: независимые проверки / Do not edit: independent checks.
import json, copy
from sklearn.decomposition import PCA as OraclePCA
from sklearn.linear_model import Ridge as OracleRidge

def check_basis():
    V=np.vstack([X[train[:5]],X[train[0]],X[train[1]]+X[train[2]]]); before=V.copy()
    Q=orthonormal_rows(V)
    assert Q.shape==(5,64)
    assert np.allclose(Q@Q.T,np.eye(5),atol=1e-8)
    assert np.allclose(V@Q.T@Q,V,atol=1e-8)
    assert np.array_equal(before,V)
    zero=V[:2]-V[:2]
    assert orthonormal_rows(zero).shape==(0,64)

def check_pca():
    for k in (3,12,64):
        A=X[train].copy(); before=A.copy(); mu,C,r=fit_pca(A,k)
        ref=OraclePCA(n_components=k,svd_solver='full').fit(A)
        assert mu.shape==(64,) and C.shape==(k,64) and r.shape==(k,)
        assert np.allclose(mu,ref.mean_)
        assert np.allclose(C.T@C,ref.components_.T@ref.components_,atol=1e-7)
        assert np.allclose(r,ref.explained_variance_ratio_,atol=1e-8)
        assert np.array_equal(A,before)

def check_project():
    ref=OraclePCA(n_components=12,svd_solver='full').fit(X[train]);mu=ref.mean_.copy();C=ref.components_.copy();A=X[valid[:7]].copy()
    before=[a.copy() for a in (A,mu,C)]
    assert np.allclose(project(A,mu,C),ref.transform(A))
    assert np.allclose(project(A[:1],mu,C),project(A,mu,C)[:1])
    assert project(A[:0],mu,C).shape==(0,12)
    assert all(np.array_equal(a,b) for a,b in zip((A,mu,C),before))

def check_ridge():
    refp=OraclePCA(n_components=12,svd_solver='full').fit(X[train]);Z=refp.transform(X[train]);labels=y[train].copy()
    for alpha in (.1,10.):
        A=Z.copy();before=A.copy();W,b=fit_ridge(A,labels,alpha)
        ref=OracleRidge(alpha=alpha).fit(A,np.eye(10)[labels])
        assert W.shape==(12,10) and b.shape==(10,)
        assert np.allclose(W,ref.coef_.T,atol=1e-7) and np.allclose(b,ref.intercept_,atol=1e-7)
        assert np.array_equal(A,before) and np.array_equal(labels,y[train])

def check_select():
    records=[]
    for k in (8,16):
        p=OraclePCA(n_components=k,svd_solver='full').fit(X[train])
        for alpha in (.1,10.):
            m=OracleRidge(alpha=alpha).fit(p.transform(X[train]),np.eye(10)[y[train]])
            records.append(dict(k=k,alpha=alpha,val_accuracy=float(np.mean(m.predict(p.transform(X[valid])).argmax(1)==y[valid]))))
    before=copy.deepcopy(records);expected=min(records,key=lambda r:(-r['val_accuracy'],r['k'],-r['alpha']))
    assert select_config(records)==expected and records==before
    a=dict(expected,k=16,alpha=.1,test_accuracy=1.0,tag='a');b=dict(expected,k=8,alpha=.1,test_accuracy=0.0,tag='b');c=dict(expected,k=8,alpha=10.,test_accuracy=0.0,tag='c')
    assert select_config([a,b,c])==c
    assert select_config([c,dict(c,tag='second')])==c

checks=[('basis',20,check_basis),('pca',25,check_pca),('project',15,check_project),('ridge',25,check_ridge),('select',15,check_select)]
score=0;details=[]
for label,points,fn in checks:
    try:
        fn();score+=points;details.append({'test':label,'points':points,'status':'PASS'})
    except Exception as exc:
        details.append({'test':label,'points':0,'status':'FAIL','error':type(exc).__name__})
print(json.dumps({'name':NAME,'group':GROUP,'assignment':'ORTH-08','score':score,'max_score':100,'details':details},ensure_ascii=False))
```
